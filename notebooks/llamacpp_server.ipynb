```python
import pickle
from dateutil import parser as dateparser
import random
```

```python
news = pickle.load(open("/exp_local/kenzosaki/data/iptc/raw_news/conflict/conflict war peace attack.pkl", "rb"))
```

```python
from datetime import datetime
from glm_based_event_analysis.graph_construction.event_component_extraction import LlamaCpp
```

```python
extractor = LlamaCpp(
    generation_params={
        "temperature": 0.0,
        "max_tokens": 512,
        "cache_prompt": True # fazer cache do system prompt
    },
    host="localhost",
    port=8080
)
#extractor._debug = True
```

```python
# sample = news[:10]
sample = random.sample(news, 30)
```

```python
titles = [item["title"] for item in sample]
descriptions = [item["description"] for item in sample]
urls = [item["url"] for item in sample]
dates = [dateparser.parse(item["published date"]) for item in sample]
```

```python
a = extractor.extract_event_components(titles, descriptions, urls, dates, n_jobs=16)
```

```python
success = sum([item.success for item in a])
success
```

```python
for title, date, extraction in zip(titles, dates, a):
    print(f"Source - Title: {title}")
    print(f"Source - Date: {date}")
    print(f"Extraction - what: {extraction.what}")
    print(f"Extraction - when: {extraction.when}")
    print()
```
