```python
import pickle
import pandas as pd
```

```python
events = pickle.load(open("../scripts/extracted_events.pkl", "rb")) # TODO: remover esse arquivo. ta com nome e lugar errado
```

```python
event_df  = pd.DataFrame(events)
```

```python
event_df.info()
```

```python
event_df.head()
```

```python
# extraindo lat e long para os primeiros paises que forem mencionados em cada evento
lats = []
longs = []
has_multiple_locations = []

for _, row in event_df.iterrows():
   
    locations = row["where"]["locations"]
    first_location = locations[0]
    
    lats.append(first_location.get("lat"))
    longs.append(first_location.get("long"))
    has_multiple_locations.append(len(locations) > 1)

# adicionando novas colunas
event_df["lat"] = lats
event_df["long"] = longs
event_df["has_multiple_locations"] = has_multiple_locations
```

```python
event_df["has_multiple_locations"].value_counts(normalize=True)
```
